## ۲. پروژهٔ نمونه: دستیار نوین‌کالا

در طول ترم اجزای یک دستیار را ساختیم. حالا آن‌ها را کنار هم می‌گذاریم:

- **RAG** (جلسه‌های ۸ و ۹) برای پرسش دربارهٔ قوانین فروشگاه
- **ایجنت با ابزار** (جلسه‌های ۱۲ و ۱۳) برای پرسش‌های باز دربارهٔ کالا، قیمت و محاسبه
- **سیستم دیالوگ** (جلسهٔ ۱۵) برای ثبت سفارش با پرسش تکمیلی و تأیید
- **مسیریاب (Router)** که هر پیام را به جزء مناسب می‌فرستد

مسیریاب همان ایدهٔ جلسهٔ ۱۳ است: پرسش‌هایی که مسیر معلومی دارند، مثل پرسش قانون یا ثبت سفارش، با جزء تخصصی و قابل‌پیش‌بینی جواب داده می‌شوند و فقط پرسش‌های باز به ایجنت می‌روند.

## ۳. ارزیابی سیستم ترکیبی

هر جزء جداگانه در جلسهٔ خودش ارزیابی شد. در سیستم ترکیبی، یک نقطهٔ شکست جدید داریم: **مسیریاب**. اگر پرسش قانون به ایجنت برود، احتمالاً باز هم جواب درست می‌گیرد، ولی کندتر و گران‌تر. اگر پیام «بله» وسط ثبت سفارش به RAG برود، گفتگو خراب می‌شود.

### دقت مسیریاب

</div>

```python
import pandas as pd

ROUTER_TESTS = [
    ("هزینه ارسال عادی چقدره؟", False, "policy"),
    ("گارانتی لپ‌تاپ چند ماهه؟", False, "policy"),
    ("قسطی می‌شه خرید کرد؟", False, "policy"),
    ("ساعت کاری پشتیبانی تا کیه؟", False, "policy"),
    ("یه گلکسی A55 می‌خوام بخرم", False, "order"),
    ("دو تا هدفون سونی سفارش بده", False, "order"),
    ("آبی", True, "order"),
    ("بله تأیید می‌کنم", True, "order"),
    ("فرق A55 و S24 FE چیه؟", False, "shopping"),
    ("سفارش NK482913 کجاست؟", False, "shopping"),
    ("دو تا A55 با ارسال سریع به تهران چقدر میشه؟", False, "shopping"),
    ("مرسی خداحافظ", False, "chitchat"),
]
rows = [{"message": m, "expected": e, "predicted": na.route(client, MODEL, m, in_order)}
        for m, in_order, e in ROUTER_TESTS]
router_eval = pd.DataFrame(rows)
print(f"router accuracy: {(router_eval.expected == router_eval.predicted).mean():.0%}")
router_eval[router_eval.expected != router_eval.predicted]
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

پیام‌های کوتاه مثل «آبی» و «بله» فقط با دانستن **وضعیت گفتگو** قابل مسیریابی‌اند. به همین دلیل `route` پارامتر `order_in_progress` را می‌گیرد. یک راه مطمئن‌تر: وقتی گفتگوی سفارش فعال است، پیام‌ها مستقیماً به `OrderBot` بروند، مگر اینکه مسیریاب با اطمینان بالا بگوید پرسش دیگری است.

## ۴. رابط کاربری با Gradio

**Gradio** با چند خط کد یک رابط وب برای نمایش می‌سازد. در Colab، گزینهٔ `share=True` یک نشانی عمومی موقت می‌دهد که می‌توانید روی گوشی یا در کلاس باز کنید. برای محصول واقعی، رابط را با FastAPI و یک فرانت‌اند واقعی می‌سازیم. این موضوع ترم ۶ است.

</div>

```python
import gradio as gr

demo_assistant = na.NovinAssistant(client, MODEL)


def respond(message, history):
    return demo_assistant.reply(message)


demo = gr.ChatInterface(
    respond,
    title="دستیار نوین‌کالا",
    description="پرسش دربارهٔ قوانین فروشگاه، کالاها و ثبت سفارش",
    examples=["هزینه ارسال سریع چقدره؟", "ارزون‌ترین لپ‌تاپ موجود چیه؟", "یه گلکسی A55 آبی می‌خوام"],
)
demo.launch(share=True)
```
