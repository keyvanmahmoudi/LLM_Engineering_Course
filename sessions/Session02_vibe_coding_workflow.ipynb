## ۱. وایب کدینگ چیست؟

اصطلاح **وایب کدینگ (Vibe Coding)** را آندری کارپاتی در فوریه ۲۰۲۵ به کار برد: برنامه‌نویس خواسته‌اش را به زبان طبیعی می‌گوید، کد را هوش مصنوعی می‌نویسد و برنامه‌نویس بیشتر نتیجه را می‌بیند تا خود کد را. این روش برای نمونه‌سازی سریع عالی است، ولی کدی که کسی نخوانده و تست نکرده برای محصول واقعی خطرناک است.

در این دوره منظور ما از وایب کدینگ نسخه حرفه‌ای آن است: **مهندسی نرم‌افزار با کمک هوش مصنوعی**. هوش مصنوعی سرعت نوشتن را چند برابر می‌کند، ولی مسئولیت درستی، امنیت و نگهداری کد همچنان با برنامه‌نویس است.

### طیف ابزارها

| سطح | نمونه ابزار | کاری که انجام می‌دهد | نقش برنامه‌نویس |
|---|---|---|---|
| تکمیل خودکار (Autocomplete) | GitHub Copilot، Continue | ادامه خط یا تابع فعلی را پیشنهاد می‌دهد | می‌نویسد و پیشنهادها را می‌پذیرد یا رد می‌کند |
| چت در ویرایشگر | Cursor Chat، Copilot Chat | به سؤال پاسخ می‌دهد و تکه کد تولید می‌کند | سؤال می‌پرسد و کد را خودش جای‌گذاری می‌کند |
| ایجنت (Agent) | Claude Code، Cursor Agent، Codex CLI، Gemini CLI | چند فایل را تغییر می‌دهد، دستور اجرا می‌کند و تست می‌گیرد | هدف و محدودیت را تعیین می‌کند و نتیجه را بازبینی می‌کند |

**نکته برای کاربران ایرانی:** بسیاری از این سرویس‌ها کاربران ایرانی را محدود می‌کنند. یک جایگزین رایگان و همیشه در دسترس، افزونه **Continue** برای VS Code است که به Ollama روی همان سیستم وصل می‌شود. کیفیتش از ابزارهای تجاری پایین‌تر است، ولی برای تکمیل خودکار و توضیح کد کاملاً کاربردی است. راهنمای نصب در انتهای نوت‌بوک آمده است.

## ۲. گردش کار حرفه‌ای

تفاوت اصلی برنامه‌نویس حرفه‌ای و تازه‌کار در استفاده از این ابزارها، **کنترل اندازه گام‌ها** است. تازه‌کار کل برنامه را یک‌جا می‌خواهد و با هزار خط کد روبه‌رو می‌شود که نمی‌فهمد. حرفه‌ای کار را به گام‌های کوچک و قابل بررسی تقسیم می‌کند:

1. **مشخصات (Spec):** چه چیزی ساخته شود، ورودی و خروجی، قواعد و موارد خاص.
2. **برنامه (Plan):** از ابزار می‌خواهیم قبل از نوشتن کد، برنامه‌اش را بگوید. برنامه را بررسی و اصلاح می‌کنیم.
3. **پیاده‌سازی گام‌به‌گام:** هر بار یک بخش کوچک.
4. **اجرا و تست:** هر گام بلافاصله اجرا و تست می‌شود.
5. **بازبینی (Review):** کد را می‌خوانیم و می‌فهمیم. کدی که نفهمیده‌ایم را قبول نمی‌کنیم.
6. **ثبت (Commit):** هر گام سالم در git ثبت می‌شود تا راه برگشت داشته باشیم.

در این جلسه همین چرخه را با مدل خودمان، به صورت خودکار، می‌سازیم تا ببینیم ابزارهای ایجنتی در درون چه کاری انجام می‌دهند.

## ۳. نوشتن مشخصات

کیفیت کد تولیدشده مستقیماً به کیفیت مشخصات بستگی دارد. مشخصات خوب این‌ها را دارد:

- امضای دقیق تابع یا رابط برنامه

بیایید تابعی بسازیم که قیمت‌های فارسی مانند «۱۲٫۵ میلیون تومان» را به عدد تبدیل کند. در فروشگاه نوین‌کالا، که در این دوره بارها با آن کار می‌کنیم، به این تابع نیاز داریم.

````python
%%writefile SPEC.md
# Spec: parse_price

Write a Python module `solution.py` with one function:

    parse_price(text: str) -> int

It converts a Persian price string to an integer amount in toman.

Rules:
- The number may use Persian (۰-۹) or English (0-9) digits.
- The decimal separator may be "." or the Persian "٫".
- Thousands separators "," or "٬" may appear and must be ignored.
- The unit word after the number is one of: "تومان" (x1), "هزار تومان" (x1,000),
  "میلیون تومان" (x1,000,000), "میلیارد تومان" (x1,000,000,000).
- The word "رایگان" (free) means 0.
- If no number can be found, raise ValueError.
- Use only the Python standard library.
````

## ۴. تست‌ها بخشی از مشخصات‌اند

بهترین راه برای اینکه به هوش مصنوعی دقیقاً بگوییم چه می‌خواهیم، **نوشتن تست قبل از کد** است. تست‌ها هم مشخصات را بدون ابهام بیان می‌کنند و هم معیار خودکاری برای پذیرش کد می‌شوند. این همان ایده **توسعه آزمون‌محور (Test-Driven Development)** است که با ابزارهای هوش مصنوعی ارزش آن چند برابر شده است.

برای تست از کتابخانه `pytest` استفاده می‌کنیم. با `parametrize` یک تست را روی چند ورودی اجرا می‌کنیم.

</div>

````python
%%writefile test_solution.py
import pytest
from solution import parse_price


@pytest.mark.parametrize("text, expected", [
    ("45000 تومان", 45_000),
    ("۴۵,۰۰۰ تومان", 45_000),
    ("۲۵۰ هزار تومان", 250_000),
    ("1.5 میلیون تومان", 1_500_000),
    ("۱۲٫۵ میلیون تومان", 12_500_000),
    ("3 میلیارد تومان", 3_000_000_000),
    ("رایگان", 0),
])
def test_parse_price(text, expected):
    assert parse_price(text) == expected


def test_no_number():
    with pytest.raises(ValueError):
        parse_price("قیمت ندارد")
````

## ۵. یک دستیار کدنویس کوچک

حالا حلقه‌ای می‌سازیم که کار ابزارهای ایجنتی را در مقیاس کوچک انجام می‌دهد:

1. مشخصات و تست‌ها را به مدل می‌دهیم و کد می‌خواهیم.
2. کد را از پاسخ استخراج و در `solution.py` ذخیره می‌کنیم.
3. تست‌ها را اجرا می‌کنیم.
4. اگر تستی شکست خورد، گزارش خطا را به مدل برمی‌گردانیم و اصلاح می‌خواهیم.
5. حداکثر سه بار تکرار می‌کنیم.

</div>

````python
import re
import subprocess

SYSTEM = (
    "You are a senior Python developer. Write clean, readable code with docstrings. "
    "Return the complete content of solution.py inside one ```python code block and nothing else."
)


def extract_code(reply):
    """Take the first fenced code block from the model's reply."""
    match = re.search(r"```(?:python)?\s*\n(.*?)```", reply, re.S)
    return match.group(1) if match else reply


def run_tests():
    result = subprocess.run(["python", "-m", "pytest", "-q", "test_solution.py"],
                            capture_output=True, text=True)
    return result.returncode == 0, result.stdout[-2500:]


def coding_loop(spec, tests, max_attempts=3):
    messages = [
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": f"{spec}\n\nThe code must pass these tests:\n\n{tests}"},
    ]
    for attempt in range(1, max_attempts + 1):
        reply = chat(client, MODEL, messages=messages, temperature=0.2)
        with open("solution.py", "w", encoding="utf-8") as f:
            f.write(extract_code(reply))
        passed, report = run_tests()
        print(f"attempt {attempt}: {report.strip().splitlines()[-1]}")
        if passed:
            return True
        messages += [
            {"role": "assistant", "content": reply},
            {"role": "user", "content": f"The tests failed:\n\n{report}\n\n"
                                        "Fix the code and return the complete file."},
        ]
    return False


spec, tests = open("SPEC.md").read(), open("test_solution.py").read()
print("all tests passed" if coding_loop(spec, tests) else "gave up after 3 attempts")
````
